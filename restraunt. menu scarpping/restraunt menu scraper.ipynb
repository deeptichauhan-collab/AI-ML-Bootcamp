```python
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

```python
url = "https://example.com/menu"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
soup = BeautifulSoup(response.text, "html.parser")
```

```python
menu = soup.find_all("div", class_="menu-item")
```

```python
dish_name = []
price = []
description = []

for item in menu:

    name = item.find("h3").text.strip()

    cost = item.find("span", class_="price").text.strip()

    desc = item.find("p").text.strip()

    dish_name.append(name)
    price.append(cost)
    description.append(desc)
```

```python
df = pd.DataFrame({
    "Dish Name": dish_name,
    "Price": price,
    "Description": description
})

df.head()
```

```python
df.to_csv("restaurant_menu.csv", index=False)

print("CSV Saved")
```
