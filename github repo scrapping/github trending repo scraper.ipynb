```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
keyword = input("Enter repository keyword (e.g., python, machine learning): ")

url = f"https://api.github.com/search/repositories?q={keyword}&sort=stars&order=desc&per_page=30"

headers = {
    "Accept": "application/vnd.github+json"
}

response = requests.get(url, headers=headers)

print(response.status_code)
```

```python
data = response.json()

len(data["items"])
```

```python
repo_name = []
owner = []
stars = []
forks = []
language = []
description = []
url_list = []

for repo in data["items"]:

    repo_name.append(repo["name"])
    owner.append(repo["owner"]["login"])
    stars.append(repo["stargazers_count"])
    forks.append(repo["forks_count"])
    language.append(repo["language"])
    description.append(repo["description"])
    url_list.append(repo["html_url"])
```

```python
df = pd.DataFrame({
    "Repository": repo_name,
    "Owner": owner,
    "Stars": stars,
    "Forks": forks,
    "Language": language,
    "Description": description,
    "GitHub URL": url_list
})

df.head()
```

```python
df.to_csv("github_repositories.csv", index=False)

print("CSV Saved Successfully")
```
